```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
data = [('Alice',25,'New York'),
        ('Bob',30,'San F'),
        ('Charlie',35,'Chicago')]

schema = '''
        name STRING,
        age INT,
        city STRING'''
df = spark.createDataFrame(data,schema = schema)
```

### Narrow Transformations

```python
df = df.filter(col('city')=='New York')
```

```python
display(df)
```

```python
df.explain()
```

### Wide Transformation

```python
data = [('Alice',25,'New York'),
        ('Bob',30,'San F'),
        ('Charlie',35,'Chicago')]

schema = '''
        name STRING,
        age INT,
        city STRING'''
df = spark.createDataFrame(data,schema = schema)
```

```python
df = df.groupBy('city').agg(max(col("age")))
```

```python
display(df)
```
