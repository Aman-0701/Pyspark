### Streaming Query

```python
my_schema = '''
                order_id INT,
                customer_id INT,
                order_date DATE,
                amount DOUBLE
    '''
```

```python
# Before running spark structure streaming, run code in batch mode to check schema or troubleshoot purpose in future

df_batch = spark.read.format('csv')\
             .option('header',True)\
             .schema(my_schema)\
             .load('/Volumes/pyspark_cata/source/db_volume/streamSource')

display(df_batch)
```

```python
df = spark.readStream.format('csv')\
          .schema(my_schema)\
          .option('header',True)\
          .load('/Volumes/pyspark_cata/source/db_volume/streamSource/')
```

### Streaming Output

```python
df.writeStream.format('delta')\
    .option('checkpointLocation','/Volumes/pyspark_cata/source/db_volume/streamSink/checkpoint')\
    .option('mergeSchema',True)\
    .trigger(once=True)\
    .start('/Volumes/pyspark_cata/source/db_volume/streamSink/data')
```
