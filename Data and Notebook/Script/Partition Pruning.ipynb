```python
data1 = [('Alice','HR',1000),('Bob','IT',2000),('Charlie','HR',3000),('David','Finance',4000),("Eva",'IT',5000),('Frank','Finance',2200)]

columns = ['name','department','salary']

df = spark.createDataFrame(data=data1, schema=columns)

output_path = '/Volumes/apachespark/csv/dataset/OutputData'

# This is called partition pruning, it creates multiple folders while storing the data
df.write\
    .mode('overwrite')\
    .partitionBy('department')\
    .parquet(output_path)

df_withpart = spark.read.format('parquet').load(output_path)
```

```python
from pyspark.sql.functions import *
```

```python
output_path_new = '/Volumes/apachespark/csv/dataset/OutputData_without_part'
df.write.mode('overwrite').parquet(output_path_new)

df_without_part = spark.read.format('parquet').load(output_path_new).filter(col('department')=='HR')
```

### Dynamic Partition Pruning

```python
df_join = df_without_part.join(df_withpart,(df_without_part['name'] == df_withpart['name']) & (df_without_part['department'] == df_withpart['department']),'inner')
```
