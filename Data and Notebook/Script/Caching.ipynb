```python
from pyspark.sql.functions import *

# Create first dataframe

data1 = [(1,'Alice'),(2,'Bob'),(3,'Charlie'),(4,'David'),(5,"Eva")]

df1 = spark.createDataFrame(data1,['id','name'])
```

```python
display(df1)
```

```python
df1 = df1.withColumn('flag',lit('Yes'))
display(df1)
```

```python
df1.cache()
```
