### Dim Customers

```python
%sql
CREATE OR REPLACE TABLE datamodelling.gold.DimCustomers  
AS
select row_number() over(order by customer_id) as DimCustomerKey, * from
(select distinct customer_id, customer_name, customer_email, customer_name_upper
from  datamodelling.silver.silver_table)
```

```python
%sql
select * from datamodelling.gold.DimCustomers
```

### Dim Products

```python
%sql
create or replace table datamodelling.gold.DimProducts
as
select row_number() over(order by product_id) as DimProductKey, * from
(select distinct product_id, product_name, product_category from datamodelling.silver.silver_table)
```

```python
%sql
select * from datamodelling.gold.DimProducts
```

### Dim Payments

```python
%sql
create or replace table datamodelling.gold.DimPaymentMethod
as
select row_number() over(order by payment_type) as DimPaymentKey, * from
(select distinct payment_type from datamodelling.silver.silver_table)
```

```python
%sql
select * from datamodelling.gold.DimPaymentMethod
```

### Dim Region

```python
%sql
create or replace table datamodelling.gold.DimRegions
as
select row_number() over(order by country) as DimRegionKey, * from
(select distinct country from datamodelling.silver.silver_table)
```

```python
%sql
CREATE OR REPLACE TABLE datamodelling.gold.DimSales AS SELECT row_number() OVER (ORDER BY order_id) as DimSaleKey, order_id, order_date, customer_id, customer_name, customer_email, product_id, product_name, product_category, payment_type, country, last_updateed, Customer_Name_Upper, processDate FROM datamodelling.silver.silver_table
```

```python
%sql
select * from datamodelling.gold.DimSales
```

### Fact Table

```python
%sql
create or replace table datamodelling.gold.FactSales
as
select 
  c.DimCustomerKey,
  p.DimProductKey,
  r.DimRegionKey,
  m.DimPaymentKey,
  s.DimSaleKey,
  f.quantity,
  f.unit_price
  from datamodelling.silver.silver_table f
  left join datamodelling.gold.DimCustomers c on f.customer_id = c.customer_id
  left join datamodelling.gold.DimProducts p on f.product_id = p.product_id
  left join datamodelling.gold.DimRegions r on f.country = r.country
  left join datamodelling.gold.DimPaymentMethod m on f.payment_type = m.payment_type
  left join datamodelling.gold.DimSales s on f.order_id = s.order_id
```

```python
%sql
select * from datamodelling.gold.FactSales
```
