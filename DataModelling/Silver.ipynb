### Merge using SQL

```python
%sql
create table if not exists datamodelling.silver.silver_table
as
select * from silver_source
```

```python
%sql
merge into datamodelling.silver.silver_table
using silver_source 
on datamodelling.silver.silver_table.order_id = silver_source.order_id
when matched then update set *
when not matched then insert *
```
